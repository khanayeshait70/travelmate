**Recommendation function**

```python
def recommend_places(
    budget,
    selected_interests,
    preferred_city=None,
    min_rating=0,
    top_n=5
):

    data = df.copy()

    # Budget filtering
    data = data[data['entry_fee'] <= budget]

    # Rating filtering
    data = data[data['rating'] >= min_rating]

    # City filtering
    if preferred_city:
        data = data[
            data['city'].str.lower() ==
            preferred_city.lower()
        ]

    # Interest matching
    selected_interests = [
        x.lower() for x in selected_interests
    ]

    data['interest_score'] = data['interest_list'].apply(
        lambda interests: (
            len(set(interests) & set(selected_interests))
            / len(set(selected_interests))
            if selected_interests else 0
        )
    )

    # Combined score
    data['final_score'] = (
        0.70 * data['interest_score'] +
        0.30 * (data['rating'] / 5)
    )

    return data.sort_values(
        'final_score',
        ascending=False
    )[
        [
            'place_name',
            'city',
            'category',
            'entry_fee',
            'rating',
            'interest_score',
            'final_score'
        ]
    ].head(top_n)
```

**Data cleaning and preprecessing**

```python
travel_df = df.copy()

# Remove duplicate places
travel_df = travel_df.drop_duplicates(subset='place_name')

# Clean text columns
text_columns = ['place_name', 'city', 'category', 'description', 'interests']

for col in text_columns:
    travel_df[col] = travel_df[col].fillna('').astype(str).str.strip()

# Fill missing numeric values
travel_df['entry_fee'] = travel_df['entry_fee'].fillna(
    travel_df['entry_fee'].median()
)

travel_df['rating'] = travel_df['rating'].fillna(
    travel_df['rating'].median()
)

print("Data Cleaning Completed!")
print(travel_df.isnull().sum())
```

```python
travel_df['interest_list'] = travel_df['interests'].str.lower().apply(
    lambda x: x.split()
)

travel_df[['place_name', 'interests', 'interest_list']].head()
```

```python
# Available interests
all_interests = sorted(set(
    interest
    for interests in travel_df['interest_list']
    for interest in interests
))

print("Available Interests:")
print(all_interests)

# Available cities
print("\nAvailable Cities:")
print(sorted(travel_df['city'].unique()))
```

```python
def recommend_places(budget, selected_interests,
                     preferred_city=None, min_rating=0, top_n=5):

    data = travel_df.copy()

    # Budget filter
    data = data[data['entry_fee'] <= budget]

    # Rating filter
    data = data[data['rating'] >= min_rating]

    # City filter
    if preferred_city and preferred_city != "All":
        data = data[
            data['city'].str.lower() == preferred_city.lower()
        ]

    # Interest matching
    selected_interests = [x.lower() for x in selected_interests]

    def calculate_interest_score(place_interests):
        if not selected_interests:
            return 0
        matched = set(place_interests) & set(selected_interests)
        return len(matched) / len(set(selected_interests))

    data['interest_score'] = data['interest_list'].apply(
        calculate_interest_score
    )

    # Rating score
    data['rating_score'] = data['rating'] / 5

    # Final recommendation score
    if selected_interests:
        data['final_score'] = (
            0.70 * data['interest_score'] +
            0.30 * data['rating_score']
        )
    else:
        data['final_score'] = data['rating_score']

    # Sort recommendations
    data = data.sort_values(
        by=['final_score', 'rating'],
        ascending=False
    )

    return data[[
        'place_name',
        'city',
        'category',
        'entry_fee',
        'rating',
        'interest_score',
        'final_score'
    ]].head(top_n)
```

**TF-IDF ANALYSIS**

```python

import pandas as pd
from pathlib import Path
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Correct CSV file path
dataset_path = Path(
    r"D:\DatascienceProject\dataset\travelmate_pakistan_tourist_places.csv"
)

# Check file
if not dataset_path.exists():
    raise FileNotFoundError(f"CSV file not found: {dataset_path}")

# Load dataset
travel_df = pd.read_csv(dataset_path)

# Combine text columns
travel_df["combined_text"] = (
    travel_df["place_name"].fillna("").astype(str) + " " +
    travel_df["category"].fillna("").astype(str) + " " +
    travel_df["description"].fillna("").astype(str) + " " +
    travel_df["interests"].fillna("").astype(str)
)

# TF-IDF Vectorization
tfidf = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2)
)

tfidf_matrix = tfidf.fit_transform(
    travel_df["combined_text"]
)

# Cosine similarity
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

print("Dataset loaded successfully!")
print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Cosine similarity shape:", cosine_sim.shape)
```

**Cosine Similarity demonstration**

```python

query = ["nature adventure mountains"]

query_vector = tfidf.transform(query)

similarity_scores = cosine_similarity(
    query_vector,
    tfidf_matrix
).flatten()

similarity_results = travel_df.copy()
similarity_results["similarity_score"] = similarity_scores

similarity_results = similarity_results.sort_values(
    "similarity_score",
    ascending=False
)

display(
    similarity_results[
        ["place_name", "city", "category", "similarity_score"]
    ].head(10)
)
```
